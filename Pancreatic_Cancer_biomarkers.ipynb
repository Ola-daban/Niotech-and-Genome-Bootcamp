```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
display(df.head())
```

```python

print("Missing Values:")
df.isnull().sum()
```

```python
median_ca19 = df['plasma_CA19_9'].median()
df['plasma_CA19_9'].fillna(median_ca19, inplace=True)
```

```python
median_ca19 = df['REG1A'].median()
df['REG1A'].fillna(median_ca19, inplace=True)
```

```python
print("After cleaning:")
print(df.isnull().sum())
```

```python
df.describe()
```

```python
df.corr(numeric_only=True)
```

```python

sns.boxplot(x='diagnosis', y='REG1B', data=df)
plt.title("Box Plot: REG1B Concentration Distribution by Diagnosis")
plt.xlabel("Diagnostic Group")
plt.ylabel("REG1B Level (ng/mL)")
plt.show()
```

```python
sns.scatterplot(x='LYVE1', y='REG1B', hue='diagnosis', data=df)
plt.title("Scatter Plot: LYVE1 vs. REG1B Biomarkers")
plt.xlabel("LYVE1 (ng/mL)")
plt.ylabel("REG1B (ng/mL)")
plt.show()
```

```python
numeric_df = df[['LYVE1', 'REG1B', 'TFF1', 'plasma_CA19_9', 'age']]

corr_matrix = numeric_df.corr()

sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Biomarker Correlation Heatmap')
plt.show()
```
